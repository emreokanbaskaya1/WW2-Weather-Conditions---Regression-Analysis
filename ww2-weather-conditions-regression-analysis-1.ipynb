```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
df = df.loc[:, df.isna().mean() < 0.70]
```

```python

df.dropna(axis=1 ,how="any", inplace=True)
```

```python
df["Date"]
```

```python
df.columns
```

```python
df["Date"] = pd.to_datetime(df["Date"],format="%Y-%m-%d")
```

```python
df.info()
```

```python
df.head()
```

```python
df["Precip"] = df["Precip"].replace("T", 0)
df["Precip"] = pd.to_numeric(df["Precip"], errors="coerce")
```

```python
df.info()
```

```python
df.corr()
```

```python
plt.figure(figsize=(12,8))
corr = df.corr()
sns.heatmap(corr, annot=True)
plt.show()
```

```python
df.columns
```

```python
cols = ["MaxTemp", "MinTemp", "MeanTemp"]
df = df[cols]
```

```python
df
```

```python
#features
X = df[["MaxTemp", "MinTemp"]]
```

```python
#target
y = df[["MeanTemp"]]
```

```python
X.head()
```

```python
y.head()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,train_size=0.20, random_state=15)
```

```python
X_train.shape
```

```python
X_test.shape
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
X_train_scaled
```

```python
X_test_scaled
```

```python
plt.subplots(figsize=(15,5))
plt.subplot(1,2,1)
sns.boxplot(data=X_train)
plt.title("X_train")
plt.subplot(1,2,2)
sns.boxplot(data=X_train_scaled)
plt.title("X_train_scaled")
plt.show()
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
```

```python
linear = LinearRegression()
linear.fit(X_train_scaled, y_train)
y_pred = linear.predict(X_test_scaled)
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
score = r2_score(y_test, y_pred)
print("Mean Absolute Error: " , mae)
print("Mean Square Error: " , mse)
print("R2 Score: " , score)
plt.scatter(y_test,y_pred)
plt.show()
```

```python
X_new = np.array([[30, 20]])
X_new_scaled = scaler.transform(X_new)
pred = linear.predict(X_new_scaled)
print("Predict MeanTemp:", pred[0][0])
```

```python
from sklearn.linear_model import Lasso
lasso = Lasso()
lasso.fit(X_train_scaled, y_train)
y_pred = lasso.predict(X_test_scaled)
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
score = r2_score(y_test, y_pred)
print("Mean Absolute Error: " , mae)
print("Mean Square Error: " , mse)
print("R2 Score: " , score)
plt.scatter(y_test,y_pred)
plt.show()
```

```python
from sklearn.linear_model import Ridge
ridge = Ridge()
ridge.fit(X_train_scaled, y_train)
y_pred = ridge.predict(X_test_scaled)
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
score = r2_score(y_test, y_pred)
print("Mean Absolute Error: ", mae)
print("Mean Squared Error: ", mse)
print("R2 Score: ", score)
plt.scatter(y_test,y_pred)
plt.xlabel("y Test")
plt.ylabel("y Pred")
plt.show()
```

```python
from sklearn.linear_model import ElasticNet
elasticnet = ElasticNet()
elasticnet.fit(X_train_scaled, y_train)
y_pred = elasticnet.predict(X_test_scaled)
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
score = r2_score(y_test, y_pred)
print("Mean Absolute Error: ", mae)
print("Mean Squared Error: ", mse)
print("R2 Score: ", score)
plt.scatter(y_test,y_pred)
plt.xlabel("y Test")
plt.ylabel("y Pred")
plt.show()
```

```python
from sklearn.linear_model import ElasticNetCV
elasticnetcv = ElasticNetCV(cv=5)
elasticnetcv.fit(X_train_scaled, y_train)
y_pred = elasticnetcv.predict(X_test_scaled)
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
score = r2_score(y_test, y_pred)
print("Mean Absolute Error: ", mae)
print("Mean Squared Error: ", mse)
print("R2 Score: ", score)
plt.scatter(y_test,y_pred)
plt.xlabel("y Test")
plt.ylabel("y Pred")
plt.show()
```
